```python
def calculaterental(cartype, days, driver=False):
    if cartype == "Economy":
        priceperday = 3000
    elif cartype == "Sedan":
        priceperday = 5000
    elif cartype == "SUV":
        priceperday = 8000
    else:
        priceperday = 0
        print("Invalid car type!")
    carcost = priceperday * days
    if driver == True:
        drivercost = 2000 * days
    else:
        drivercost = 0
    if days >= 7:
        discountpercent = 15
    elif days >= 3:
        discountpercent = 10
    else:
        discountpercent = 0
    discountamount = (carcost * discountpercent) / 100
    finalbill = carcost - discountamount + drivercost
    print("Car Type =", cartype)
    print("Days =", days)
    print("Driver Required =", driver)
    print("Car Cost =", carcost)
    print("Driver Cost =", drivercost)
    print("Discount (", discountpercent, "%) =", discountamount)
    print("Final Bill =", finalbill)
    print("------------------------")
    return finalbill
calculaterental("Economy", 2)
calculaterental("Sedan", 5, driver=True)
calculaterental("SUV", 8, driver=True)
calculaterental("Sedan", 4)
```
